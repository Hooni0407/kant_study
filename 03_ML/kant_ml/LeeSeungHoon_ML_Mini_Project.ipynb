```python
# M1
# ==!!!============================================
plt.rcParams["font.family"] = "Malgun Gothic"

df = sns.load_dataset("diamonds")
df.to_csv("./datasets/office.csv", index=False)

print("타입")
print(df.shape)

# 자료형 조회
print("자료형 조회")
df.info()

# 빈칸
print("빈칸 합계")
df.isnull().sum()

# 기초 통계량
print("기초 통계량")
df.describe()

# 생각해볼 질문
# - 평균값이 중앙값보다 극단적으로 높은 상태이다.
```

```python
# M2
x = (df["x"] == 0).sum()
print(f"x가 0인 값의 갯수:{x}개")
y = (df["y"] == 0).sum()
print(f"y가 0인 값의 갯수:{y}개")
z = (df["z"] == 0).sum()
print(f"z가 0인 값의 갯수:{z}개")

print("="*40)

zero_rows = df[(df["x"] == 0) | (df["y"] == 0) | (df["z"] == 0)]
print(f"0이 하나라도 있는 행이 총 {len(zero_rows)}개")
zero_rows

print("="*40)

outlier = df[(df["y"] >= 20) | (df["z"] >= 20)]
print(f"y나 z가 20mm를 넘는 행 : {outlier}")
outlier

print("="*40)

clean = df[(df["x"] != 0) & (df["y"] != 0) & (df["z"] != 0) & (df["y"] < 20) & (df["z"] < 20)]

print(f"제거 후 행 수 : {len(df) - len(clean)}")
```
