```python
import numpy as np
import pandas as pd

title = ["봄바람", "Run Up", "새벽 감성", "Neon City", "첫눈", "Jump!", "바다소리", "Fire Dance"]

genre = ["발라드", "댄스", "발라드", "힙합", "발라드", "댄스", "인디", "댄스"]

plays = [12, 35, 8, 22, 15, 28, 5, 19]

playlist = pd.DataFrame({
    "title" : title,
    "genre" : genre,
    "plays" : plays
})

playlist
```

```python
# 이번달 총 재생 수
print(playlist["plays"].sum())
```

```python
# 재생 수 20 이상인 곡
for t, g, p in zip(title, genre, plays):
    if p >= 20: 
        print(f"title : {t} | genre : {g} | plays : {p}")
```

```python
top3 = playlist.sort_values("plays", ascending=False).head(3)
top3
```

```python
playlist.groupby("genre")["plays"].sum().sort_values(ascending=False)
```
