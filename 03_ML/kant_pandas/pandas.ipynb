```python
amounts = [3200, 1500, 15000, 6500, 32000, 4800,
           10900, 1500, 2700, 3000, 18500, 7000]

print(amounts)
print("합계 : ", sum(amounts))

print(amounts * 2)
```

```python
import numpy as np

arr = np.array(amounts)
arr * 2

print("합계 : ", arr.sum())
print("평균 : ", arr.mean())
print("최소 : ", arr.min())
print("최대 : ", arr.max())

print("모양 : ", arr.shape, "데이터 타입 : ", arr.dtype) # (12,)라는 건 1차원 배열이란 뜻
```

```python
print(arr >= 10000)
arr[arr >= 10000]
```
